```python
# printing numerical columns only
numerical_data = data.select_dtypes(include=[np.number])

# Print column names
print(list(numerical_data.columns))
```

```python
#  printing categorical columns only
categorical_cols = data.select_dtypes(include=['object', 'category']).columns.tolist()
print("Categorical columns:", categorical_cols)
```

```python
# Univariate Analysis, Numerical Variables
# Distribution of numerical variables
numerical_cols = ['age', 'monthly_income', 'monthly_expenses', 'monthly_savings', 'credit_score', 'debt_to_income_ratio', 'financial_health_score']

fig, axes = plt.subplots(3, 3, figsize=(15, 12))
axes = axes.ravel()
for i, col in enumerate(numerical_cols):
    if i < len(axes):
        data[col].hist(ax=axes[i], bins=30, color='skyblue', edgecolor='black')
        axes[i].set_title(f'Distribution of {col}')
        axes[i].set_xlabel(col)
        axes[i].set_ylabel('Frequency')
plt.tight_layout()
plt.show()
```

```python
# Categorical Variables
# Distribution of categorical variables
categorical_cols = ['employment_type', 'risk_category']
fig, axes = plt.subplots(1, 2, figsize=(12, 5))
for i, col in enumerate(categorical_cols):
    data[col].value_counts().plot(kind='bar', ax=axes[i], color='lightgreen')
    axes[i].set_title(f'Distribution of {col}')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
# Multivariate Analysis, Numerical vs Categorical
# Numerical variables vs categorical variables
fig, axes = plt.subplots(2, 2, figsize=(15, 10))

# Financial health score vs employment type
sns.boxplot(x='employment_type', y='financial_health_score', data=data, ax=axes[0,0])
axes[0,0].set_title('Financial Health Score vs Employment Type')

# Credit score vs Risk Category
sns.boxplot(x='risk_category', y='credit_score', data=data, ax=axes[0,1])
axes[0,1].set_title('Credit Score vs Risk Category')

# Debt to income ratio vs Employment Type
sns.boxplot(x='employment_type', y='debt_to_income_ratio', data=data, ax=axes[1,0])
axes[1,0].set_title('Debt to Income Ratio vs Employment Type')

# Monthly savings vs Risk Category
sns.boxplot(x='risk_category', y='monthly_savings', data=data, ax=axes[1,1])
axes[1,1].set_title('Monthly Savings vs Risk Category')

plt.tight_layout()
plt.show()
```

```python
# Correlation Heatmap (Numerical Variables Only)
# Correlation matrix for numerical variables
numerical_cols = ['age', 'monthly_income', 'monthly_expenses', 'monthly_savings', 'credit_score', 'debt_to_income_ratio', 'financial_health_score']
correlation_matrix = data[numerical_cols].corr()

plt.figure(figsize=(10, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', center=0)
plt.title('Correlation Matrix of Numerical Variables')
plt.show()
```
